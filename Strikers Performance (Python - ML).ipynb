```python
#APPLYING DATA TRANSFORMATION TECHNIQUES (yeojohnson transformation)
```

```python
from scipy.stats import yeojohnson
```

```python
def yeojohnson_transformation(data, column):
    transformed_data, lamda = yeojohnson(data[column])
    data[f'{column}_yeojohnson'] = transformed_data
    stat, p_value = shapiro(data[f'{column}_yeojohnson'])
    fig, axes = plt.subplots(1,1, figsize = (2,2))
    sns.kdeplot(data[f'{column}_yeojohnson'])
    print(lamda)
    print(p_value)
```

```python
data['Goals Scored'] = yeojohnson_transformation(data, 'Goals Scored')
```

```python
data['Big Game Performance'] = yeojohnson_transformation(data, 'Big Game Performance')
```

```python
data['Penalty Success Rate'] = yeojohnson_transformation(data, 'Penalty Success Rate')
```

```python
data['Aerial Duels Won'] = yeojohnson_transformation(data, 'Aerial Duels Won')
```

```python
from scipy import stats
```

```python
#1. CHECKING SIGNIFICANT DIFFERENCE BETWEEN THE TOTAL SCOES OF PLAYER (NATIONALITY WISE):

spain = data.query('Nationality == "Spain"')['Total_Scores']
brazil = data.query('Nationality == "Brazil"')['Total_Scores']
france = data.query('Nationality == "France"')['Total_Scores']
germany = data.query('Nationality == "Germany"')['Total_Scores']
england = data.query('Nationality == "England"')['Total_Scores']

t_stat, p_value = stats.f_oneway(spain, brazil, france, germany, england)
print("p_value : ", round(p_value,2))
```

```python
#2. CHECKING SIGNIFICANT DIFFERENCE BETWEEN THE TOTAL SCOES OF PLAYER (FOOTEDNESS WISE):

left = data.query('Footedness == "Left-footed"')['Total_Scores']
right = data.query('Footedness == "Right-footed"')['Total_Scores']

t_stat, p_value = stats.ttest_ind(left, right)
print("p_value : ", p_value)
```

```python
from sklearn.decomposition import PCA
```

```python
#APPLY PCA:

pca = PCA()
dimensionality_reduced_data = pca.fit_transform(scaled_data)
```

```python
#VARIANCE EXPLAINED RATIO PLOT:

var_ratio = pca.explained_variance_ratio_
cum_var = np.cumsum(var_ratio)

plt.figure(figsize=(4,4))
plt.plot(cum_var, marker='o')
plt.xlabel("No of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA - Variance Explained Ratio")
plt.grid(True)
plt.show()
```

```python
# Saare features lene par (15 components) 100% variance (information) de raha hai par 10 components lene par 80% variance de raha hai.
# Yahan elbow point nahibhai curve mein kyunki har feature equal amount of variance de raha hai.
```

```python
#Either take 15 components -> give 100% information
#Or take 10 components -> give 80% information -> good balance between dimensionality reduction and information retained
```

```python
#SELECT NUMBER OF COMPONENTS: n = 15

pca = PCA(n_components = 15)
dimensionality_reduced_data = pca.fit_transform(scaled_data)
```

```python
#DATA MINING: K-MEANS CLUSTERING METHOD
```

```python
from sklearn.cluster import KMeans
```

```python
#CALCULATE WCSS

wcss = []
for K in range(1,11):
    kmeans = KMeans(n_clusters = K, random_state=42)
    kmeans.fit(dimensionality_reduced_data)
    wcss_values = kmeans.inertia_
    wcss.append(wcss_values)
```

```python
#SELECTING OPTIMAL NUMBER OF CLUSTERS: ELBOW METHOD

plt.figure(figsize=(4,4))
plt.plot(range(1,11), wcss, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('wcss')
plt.title('Elbow Method')
plt.grid(True)
plt.show()
```

```python
#FINAL MODEL WITH K = 4

kmeans = KMeans(n_clusters = 4, random_state = 42)
Clusters = kmeans.fit_predict(dimensionality_reduced_data)
```

```python
#CALLING OUT THE CLUSTERED DATA OUTPUT:STORED IN VARIABLE final_data
Clusters
```

```python
#ADDING CLUSTERED OUTPUT TO THE DATASET

clustered_data = pd.DataFrame(dimensionality_reduced_data, columns=[f'PC{K+1}' for K in range(dimensionality_reduced_data.shape[1])])
clustered_data['Clustered_Performance'] = Clusters
```

```python
clustered_data.head()
```

```python
#GIVE THE CLUSTERS, CATEGORICAL LABELS TO IDENTIFY THE STRIKERS PERFORMANCE AS top-performing or worst-performing
```

```python
#CHECK WHICH CLUSTER IS GOOD AT WHICH PARTICULAR FEATURE: CHECK CLUSTER CENTROIDS -> Give avg value of each feature for each cluster

print("Cluster centers (PCA space):\n", kmeans.cluster_centers_)
```

```python
#CLUSTER RANKING: CALCULATE Total Scores AND CHECK WHICH CLUSTER HAS Total Scores AND RATE ACCORDINGLY (MAPPING)

clustered_data['Total_Performance_Scores'] = clustered_data['PC1'] + clustered_data['PC2'] + clustered_data['PC3'] + clustered_data['PC4'] + clustered_data['PC5'] + clustered_data['PC6'] + clustered_data['PC7'] + clustered_data['PC8'] + clustered_data['PC9'] + clustered_data['PC10'] + clustered_data['PC11'] + clustered_data['PC12'] + clustered_data['PC13'] + clustered_data['PC14'] + clustered_data['PC15']
```

```python
clusters_ranking = clustered_data.groupby('Clustered_Performance')['Total_Performance_Scores'].mean().sort_values(ascending=False)
print(clusters_ranking)
```

```python
#MAPPING

mapping = {2 : 'Top-Performing Strikers', 3 : '2nd Top-Performing Strikers', 1 : '3rd Top-Performing Strikers', 0 : 'Last Top Performing Strikers'}
clustered_data['Strikers_Performance'] = clustered_data['Clustered_Performance'].map(mapping)
```

```python
#SILHOUETTE SCORE:

from sklearn.metrics import silhouette_score

score = silhouette_score(dimensionality_reduced_data, Clusters)
print("Silhouette Score:", score)
```

```python
#DAVIES-BOULDIN INDEX:

from sklearn.metrics import davies_bouldin_score
dbi = davies_bouldin_score(dimensionality_reduced_data, Clusters)
print("Davies-Bouldin Index:", dbi)
```
